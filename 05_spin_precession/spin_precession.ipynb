
## What “spin” means (and doesn’t)
- Not a tiny ball spinning: spin is **intrinsic** angular momentum.
- It behaves like angular momentum: it has a magnitude and direction, precesses in a magnetic field, and carries a magnetic moment $\boldsymbol\mu$.

## Basic quantum numbers
- A nucleus has total spin $I$ (sometimes written $S$). Allowed projections are $m_I=-I,-I+1,\dots,+I$.
- Fixed magnitude:
$$
|\mathbf I|=\sqrt{I(I+1)}\,\hbar .
$$
- Only the component along a chosen axis (usually the field) is quantized: $I_z = m_I\hbar$.
- Magnetic moment proportional to spin:
$$
\boldsymbol\mu=\gamma\,\mathbf I
\qquad\text{(equivalently } \boldsymbol\mu = g\,\frac{q}{2m}\,\mathbf I\text{)}.
$$

## Where nuclear spin comes from
- Protons and neutrons each have spin $1/2$ and may carry orbital angular momentum inside the nucleus; the total $I$ is the vector sum.
- Even–even nuclei $\rightarrow I=0$ (paired spins).  
  Odd-$A$ nuclei typically have $I\neq 0$ (e.g., $1/2,3/2,5/2,\dots$).
- Integer $I$ $\rightarrow$ boson; half-integer $I$ $\rightarrow$ fermion.

## Spin in a magnetic field: precession and energy levels
- Torque in a uniform field $\mathbf B$: $\boldsymbol\tau=\boldsymbol\mu\times\mathbf B$.
- **Larmor precession**:
$$
\frac{d\mathbf I}{dt}=\gamma\,\mathbf I\times\mathbf B
\quad\Rightarrow\quad
\omega_L=|\gamma|\,B .
$$
- **Zeeman energies**:
$$
E=-\boldsymbol\mu\cdot\mathbf B=-\gamma\,\hbar\,m_I\,B ,
$$
so the $2I+1$ levels are equally spaced by
$$
\Delta E=\hbar\omega_L=\hbar|\gamma|B .
$$

## Why NMR/MRI use radio waves
- Nuclear moments are small, so $\Delta E$ is tiny. For protons,
$$
\frac{\gamma}{2\pi}\approx 42.58~\text{MHz/T},\qquad
f=\frac{\omega_L}{2\pi}=\frac{|\gamma|B}{2\pi} .
$$

## “Alignment” and relaxation
- At thermal equilibrium, slightly more spins occupy the lower-energy state (moment roughly along $\mathbf B$); net polarization is small at room temperature.
- Relaxation governs dynamics:
  - $T_1$ (spin–lattice): populations return to equilibrium.
  - $T_2$ (spin–spin): transverse coherence dephases.

## Extra: quadrupole moments
- Nuclei with $I\ge 1$ have an electric quadrupole moment that couples to electric-field gradients (NQR, line broadening).  
  Spin-$1/2$ nuclei (e.g., $^1\text{H}$, $^{13}\text{C}$, $^{31}\text{P}$) do not.


# Angular momentum, spin, magnetic moments, and polarization

**Total angular momentum.** We write $ \mathbf{J} = \mathbf{L} + \mathbf{S} $. Here $ \mathbf{L} $ is orbital and $ \mathbf{S} $ is intrinsic (spin). In the present **nuclear** context we focus on spin: the nucleus as a whole is (effectively) a point-like particle in an external field, so orbital motion about the lab is irrelevant; the magnetic response is dominated by intrinsic spin.\*
  
\*In detailed nuclear-structure models, individual nucleons have orbital and spin contributions that combine to a total nuclear spin $ \mathbf{I} $, but for NMR-style discussions of a nucleus in a field, we treat the observable moment as arising from intrinsic spin.

**Spin operators.** Components of $ \mathbf{S} $ satisfy the usual angular-momentum algebra $ [S_i, S_j] = i\hbar\,\varepsilon_{ijk}\,S_k $. The eigenvalues are
$$
S^2 \ket{s,m_s} = s(s+1)\hbar^2 \ket{s,m_s},\qquad 
S_z \ket{s,m_s} = m_s\hbar \ket{s,m_s},
$$
with $ s = 0, \tfrac12, 1, \tfrac32, \dots $ and $ m_s = -s, -s+1, \dots, s $. (You may also see $h$ in place of $\hbar$; for angular momentum, $ \hbar \equiv h/2\pi $ is the natural constant.)

**What does “spin-$\tfrac12$” mean?** A spin-$\tfrac12$ object has two allowed $S_z$ outcomes on measurement: $ m_s=\pm \tfrac12 $. The state space transforms under the **fundamental representation of SU(2)**, which is why the system is intrinsically two-level and exhibits phenomena like a $4\pi$ (not $2\pi$) sign periodicity at the level of state vectors. Importantly, this is **not** a tiny classical ball physically spinning; “spin” is an intrinsic quantum degree of freedom that carries angular momentum $ \tfrac{\hbar}{2} $.

**Magnetic moment from spin.** The magnetic dipole moment is proportional to spin,
$$
\boldsymbol{\mu} = \gamma\,\mathbf{S},
$$
where $ \gamma $ is the **gyromagnetic ratio** (a constant that encodes charge, mass, and $g$-factor). For a particle with charge $q$ and mass $m$ one often writes $ \gamma = g\,\dfrac{q}{2m} $. For electrons, $ \gamma < 0 $ (because $ q=-e $); for many nuclei, $ \gamma > 0 $. Two common magnetons are $ \mu_B = \dfrac{e\hbar}{2m_e} $ (Bohr) and $ \mu_N = \dfrac{e\hbar}{2m_p} $ (nuclear).

**Zeeman energies and alignment.** In a static magnetic field $ \mathbf{B} = B\,\hat{\mathbf{z}} $,
$$
E = -\boldsymbol{\mu}\!\cdot\!\mathbf{B} = -\gamma\,\mathbf{S}\!\cdot\!\mathbf{B} = -\gamma\,S_z\,B.
$$
For spin-$\tfrac12$ this gives two levels
$$
E_{m_s=\pm\tfrac12} = \mp \tfrac12\,\hbar\gamma\,B,
$$
with splitting $ \Delta E = \hbar\gamma B $. The lower-energy orientation depends on the **sign** of $ \gamma $: if $ \gamma>0 $ the $ m_s=+\tfrac12 $ (moment parallel to $ \mathbf{B} $) is lower; if $ \gamma<0 $ the $ m_s=-\tfrac12 $ is lower. Coherently, spins **precess** at the Larmor frequency $ \omega_0 = \gamma B $.

**Polarization and magnetization.** For an ensemble, define the polarization
$$
P \equiv \frac{N_\uparrow - N_\downarrow}{N_\uparrow + N_\downarrow},
$$
which relates to magnetization $ \mathbf{M} = n\,\boldsymbol{\mu}\,P\,\hat{\mathbf{z}} $ ($n$ = number density). In thermal equilibrium for noninteracting spins,
$$
P_{\rm eq}(B,T) = \tanh\!\left(\frac{\mu\,B}{k_B T}\right),
$$
so $P$ grows with $|B|$ and decreases with $T$. (For nuclei, $ \mu \sim \mu_N $ is tiny, so $P$ is usually small at room temperature; for electrons it is larger.)

```python
%matplotlib inline
import matplotlib.pyplot as plt 
import numpy as np 
from ipywidgets import interact, IntSlider, FloatSlider
from IPython.display import clear_output, HTML, display
import os

# --- Interactive visualization of spins, magnetization, and Zeeman splitting ---
# The depiction is qualitative (not an exact simulation).

# Lattice size and a fixed latent randomness so spins "flip" monotonically with B
N = 18
rng = np.random.default_rng(12345)
latent = rng.random((N, N))  # each site has a fixed threshold in [0,1)

# Controls how sharply spins align with B (bigger -> sharper flips)
alpha = 1.6  # arbitrary

def spin_config(B):
    """
    Return an N x N array of spins (+1 for 'up', -1 for 'down') as a function of field B.
    We use a smooth monotonic mapping p_up(B) so that increasing |B| flips more spins.
    """
    p_up = 0.5 * (1.0 + np.tanh(alpha * B))  # in [0,1]
    spins = np.where(latent < p_up, 1, -1)
    return spins

def zeeman_energies(B, gamma=1.0, hbar=1.0):
    """
    Zeeman energies for a spin-1/2 in a field along z, using E = -gamma * S_z * B and S_z = m_s * hbar.
    We set hbar=1 for plotting (arbitrary units). Energies are [E_{m_s=+1/2}, E_{m_s=-1/2}].
    """
    E_plus = -gamma * ( +0.5 * hbar) * B
    E_minus = -gamma * ( -0.5 * hbar) * B
    return E_plus, E_minus

def view(B=0.0):
    """Redraw all three visuals for the given B: lattice, magnetization, Zeeman splitting."""
    spins = spin_config(B)
    P = spins.mean()  # polarization in [-1, 1]

    clear_output(wait=True)
    display(HTML(f"<h3>Spin lattice (N={N}×{N}), B = {B:.2f} (arb. units)</h3>"))

    # --- Figure 1: Spin lattice ---
    plt.figure(figsize=(6, 6))
    y, x = np.indices(spins.shape)
    up_y, up_x = np.where(spins > 0)
    dn_y, dn_x = np.where(spins < 0)
    # Up = triangle up; Down = triangle down
    plt.scatter(up_x, up_y, marker='^', s=36)
    plt.scatter(dn_x, dn_y, marker='v', s=36)
    plt.gca().invert_yaxis()
    plt.gca().set_aspect('equal', 'box')
    plt.xticks([]); plt.yticks([])
    plt.title("Spins (▲ up, ▼ down)")
    plt.show()

    # --- Figure 2: Magnetization / polarization vector ---
    display(HTML("<h3>Magnetization / Polarization</h3>"))
    plt.figure(figsize=(6, 2.4))
    plt.xlim(0, 1)
    plt.ylim(-1, 1)
    plt.axhline(0)
    # Draw an arrow whose length scales with |P| and direction with sign(P)
    arrow_len = 0.75 * abs(P)
    # Use small positive length to keep head visible at P=0
    plt.arrow(0.1, 0.0, max(arrow_len, 0.0001), 0.0, head_width=0.06, head_length=0.03, length_includes_head=True)
    # Annotate sign
    sign_str = "↑ (parallel)" if P >= 0 else "↓ (antiparallel)"
    plt.text(0.92, 0.0, f"P = {P:.2f}  {sign_str}", ha='right', va='center', fontsize=11)
    plt.title("Polarization grows in magnitude with |B| (qualitative)")
    plt.axis('off')
    plt.show()

    # --- Figure 3: Zeeman splitting for a spin-1/2 ---
    display(HTML("<h3>Zeeman splitting for spin-1/2</h3>"))
    E_plus, E_minus = zeeman_energies(B, gamma=1.0, hbar=1.0)  # arbitrary units
    E_vals = [E_plus, E_minus]
    labels = [r"$m_s=+\frac{1}{2}$", r"$m_s=-\frac{1}{2}$"]

    plt.figure(figsize=(6, 3.2))
    pad = 0.6
    ymin = min(E_vals) - pad
    ymax = max(E_vals) + pad
    plt.ylim(ymin, ymax)
    plt.xlim(0, 1)
    for e, lab in zip(E_vals, labels):
        plt.hlines(e, 0.15, 0.85)
        plt.text(0.88, e, lab, va='center', fontsize=11)
    # Annotate splitting ΔE = ħγ|B| (ħ=1 here)
    plt.text(0.50, 0.5*(E_plus+E_minus), r"$\Delta E = \hbar\,\gamma\,|B|$", ha='center', va='center', fontsize=12)
    plt.title("Two levels split linearly with B (sign of γ sets which is lower)")
    plt.axis('off')
    plt.show()

# Interactive slider for the magnetic field B (arbitrary units, negative allowed)
interact(view, B=FloatSlider(value=0.0, min=-2.0, max=2.0, step=0.05, description="B"));
```

## Precession: $\dfrac{d\boldsymbol{\mu}}{dt} = \gamma\,\boldsymbol{\mu} \times \mathbf{B}$

A magnetic moment in a static field experiences a torque
$$
\boldsymbol{\tau} = \boldsymbol{\mu} \times \mathbf{B},
$$
and because $ \boldsymbol{\mu} = \gamma\,\mathbf{S} $, the spin’s equation of motion is
$$
\frac{d\boldsymbol{\mu}}{dt} = \gamma\,\boldsymbol{\mu} \times \mathbf{B}.
$$
This is **Larmor precession**. The cross product ensures the time derivative is perpendicular to $ \boldsymbol{\mu} $, so $ |\boldsymbol{\mu}| $ is constant while its **direction** precesses around $ \mathbf{B} $ at angular frequency
$$
\omega_0 = \gamma B.
$$
The sign of $ \gamma $ sets the **sense** (clockwise vs. counterclockwise) of precession. If $ \mathbf{B} = B\,\hat{\mathbf{z}} $, then the longitudinal component is conserved, $ \mu_z(t)=\mu_z(0) $, and the transverse part rotates:
$$
\boldsymbol{\mu}_\perp(t) = \boldsymbol{\mu}_\perp(0)\cos(\omega_0 t) + \hat{\mathbf{z}} \times \boldsymbol{\mu}_\perp(0)\,\sin(\omega_0 t).
$$

### Spinning-top analogy
Think of a classical **spinning top** with angular momentum $ \mathbf{L} $ in gravity. The weight applies a torque $ \boldsymbol{\tau} $ that is **perpendicular** to $ \mathbf{L} $, so
$$
\frac{d\mathbf{L}}{dt} = \boldsymbol{\tau}
$$
does not (initially) tip the top over; it makes $ \mathbf{L} $ **precess** about the vertical at some rate $ \Omega $ while $ |\mathbf{L}| $ remains (approximately) fixed. In the magnetic case, the role of gravity’s torque is played by $ \boldsymbol{\tau} = \boldsymbol{\mu}\times\mathbf{B} $, and $ \mathbf{L} $ is replaced by $ \boldsymbol{\mu} $ (or equivalently $ \mathbf{S} $). The result is the same geometry: the tip of $ \boldsymbol{\mu} $ traces out a **cone** around $ \mathbf{B} $ at rate $ \omega_0 = \gamma B $.

In real spin ensembles (e.g., NMR/ESR), precession is the torque-only part of the **Bloch equations**; additional relaxation processes (characterized by $ T_1 $ and $ T_2 $) cause longitudinal recovery and transverse dephasing, respectively. Here we focus on the ideal, torque-only precession to build geometric intuition.

```python
%matplotlib inline
import matplotlib.pyplot as plt 
import numpy as np 
from ipywidgets import interact, IntSlider, FloatSlider
from IPython.display import clear_output, HTML, display
import os

# --- helpers ---
def set_equal_3d(ax, X, Y, Z, pad=0.20):
    x_min, x_max = np.min(X), np.max(X)
    y_min, y_max = np.min(Y), np.max(Y)
    z_min, z_max = np.min(Z), np.max(Z)
    xr, yr, zr = x_max - x_min, y_max - y_min, z_max - z_min
    r = max(xr, yr, zr) or 1.0
    cx, cy, cz = (x_max + x_min)/2, (y_max + y_min)/2, (z_max + z_min)/2
    h = (1 + pad) * r / 2
    ax.set_xlim(cx - h, cx + h)
    ax.set_ylim(cy - h, cy + h)
    ax.set_zlim(cz - h, cz + h)

# Make the figure/axis once
fig = plt.figure(figsize=(7.5, 7.5))
ax = fig.add_subplot(111, projection='3d')
ax.view_init(elev=22, azim=-60)

def draw(phi_deg=25.0, theta_deg=35.0, Jmag=1.8, torque_scale=0.25):
    # Re-render the figure on every slider change (needed for %matplotlib inline)
    clear_output(wait=True)

    ax.cla()
    ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')
    ax.view_init(elev=22, azim=-60)

    phi   = np.deg2rad(phi_deg)
    theta = np.deg2rad(theta_deg)

    # Angular momentum tip
    tip = np.array([
        Jmag*np.sin(theta)*np.cos(phi),
        Jmag*np.sin(theta)*np.sin(phi),
        Jmag*np.cos(theta)
    ])

    # Precession circle
    t = np.linspace(0, 2*np.pi, 400)
    circ_x = Jmag*np.sin(theta)*np.cos(t)
    circ_y = Jmag*np.sin(theta)*np.sin(t)
    circ_z = np.full_like(t, Jmag*np.cos(theta))
    ax.plot(circ_x, circ_y, circ_z, lw=2.0)

    # Direction arrows along the circle (+phi direction)
    for a in np.deg2rad([0, 120, 240]):
        pt = np.array([
            Jmag*np.sin(theta)*np.cos(a),
            Jmag*np.sin(theta)*np.sin(a),
            Jmag*np.cos(theta)
        ])
        tan = np.array([-np.sin(a), np.cos(a), 0.0])
        ax.quiver(pt[0], pt[1], pt[2],
                  0.18*Jmag*tan[0], 0.18*Jmag*tan[1], 0.18*Jmag*tan[2],
                  arrow_length_ratio=0.4, linewidth=1.2)

    # J vector
    ax.quiver(0, 0, 0, tip[0], tip[1], tip[2], arrow_length_ratio=0.06, linewidth=2.4)
    ax.text(tip[0]*1.02, tip[1]*1.02, tip[2]*1.02, 'J', fontsize=12)

    # dJ = N·dt at the tip (tangent)
    tan_tip = np.array([-np.sin(phi), np.cos(phi), 0.0])
    tan_tip /= (np.linalg.norm(tan_tip) + 1e-12)
    dJ = torque_scale * tan_tip
    ax.quiver(tip[0], tip[1], tip[2], dJ[0], dJ[1], dJ[2],
              arrow_length_ratio=0.4, linewidth=2.2)
    ax.text(tip[0] + 1.15*dJ[0], tip[1] + 1.15*dJ[1], tip[2] + 1.15*dJ[2],
            'dJ = N·dt', fontsize=10)

    # z-axis
    zmax = max(Jmag*1.2, 1.2)
    ax.quiver(0, 0, 0, 0, 0, zmax, arrow_length_ratio=0.05, linewidth=1.8)
    ax.text(0, 0, zmax*1.04, 'z', fontsize=12)

    # gravity (down)
    g_len = 0.6*zmax
    ax.quiver(0.1*zmax, -0.1*zmax, 0.95*zmax, 0, 0, -g_len,
              arrow_length_ratio=0.08, linewidth=1.8)
    ax.text(0.1*zmax, -0.1*zmax, 0.95*zmax - g_len - 0.05*zmax, 'g (down)', fontsize=10)

    # translucent cone
    r = np.linspace(0, Jmag*np.sin(theta), 24)
    ang = np.linspace(0, 2*np.pi, 72)
    R, A = np.meshgrid(r, ang)
    Xc = R*np.cos(A); Yc = R*np.sin(A)
    Zc = (Jmag*np.cos(theta) / (Jmag*np.sin(theta) + 1e-9)) * R
    ax.plot_surface(Xc, Yc, Zc, rstride=1, cstride=1, alpha=0.08, linewidth=0, antialiased=True)

    # θ arc
    arc_r = 0.35*Jmag
    tt = np.linspace(0, theta, 60)
    ax.plot(arc_r*np.sin(tt)*np.cos(phi),
            arc_r*np.sin(tt)*np.sin(phi),
            arc_r*np.cos(tt), lw=1.8)
    k = len(tt)//2
    ax.text(arc_r*np.sin(tt[k])*np.cos(phi)*1.04,
            arc_r*np.sin(tt[k])*np.sin(phi)*1.04,
            arc_r*np.cos(tt[k])*1.04, 'θ', fontsize=12)

    # bounds
    X = np.concatenate([circ_x, [0, 0, tip[0]], [tip[0] + dJ[0]]])
    Y = np.concatenate([circ_y, [0, 0, tip[1]], [tip[1] + dJ[1]]])
    Z = np.concatenate([circ_z, [0, zmax, tip[2]], [tip[2] + dJ[2]]])
    set_equal_3d(ax, X, Y, Z, pad=0.35)

    ax.set_title('Precession of Angular Momentum under Gravity: J precesses about +z', pad=12)

    # Re-display the updated figure
    display(fig)

# Sliders (using only your imports)
interact(
    draw,
    phi_deg=FloatSlider(min=0, max=360, step=1, value=25, description='φ (deg)'),
    theta_deg=FloatSlider(min=0.1, max=85, step=0.1, value=35, description='θ (deg)'),
    Jmag=FloatSlider(min=0.8, max=2.5, step=0.05, value=1.8, description='|J|'),
    torque_scale=FloatSlider(min=0.05, max=0.6, step=0.01, value=0.25, description='Torque')
);
```
