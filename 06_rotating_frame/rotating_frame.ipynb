# Rotating Frame 

We use only a **passive** change of coordinates (rotate the axes, not the magnetization). 
---

## 1) Circularly polarized RF in the lab frame

We model the RF drive as a **circularly polarized** field precessing about $\hat z$ at angular frequency $\omega_{\mathrm{rf}}$ with a phase offset $\phi$, plus a static field $B_0\hat z$:
$$
\mathbf B_{\text{lab}}(t)
=
\begin{bmatrix}
B_1\cos(\omega_{\mathrm{rf}}t+\phi)\\\\
B_1\sin(\omega_{\mathrm{rf}}t+\phi)\\\\
B_0
\end{bmatrix}.
$$

Here $B_1$ is the (constant) RF amplitude during the pulse; $B_0$ the main field; and $\phi$ sets the transverse orientation of the RF at $t=0$.

---

## 2) Passive rotating-frame components

We introduce a rotating coordinate system whose axes spin **about $\hat z$** at angular speed $\omega_{\mathrm{ref}}$.
A passive change of coordinates means: *rotate the axes by $R_z(\omega_{\mathrm{ref}} t)$; keep the physical vector the same*.
Components in the primed frame are therefore
$$
\mathbf B'(t) \;=\; R_z(-\omega_{\mathrm{ref}} t)\,\mathbf B_{\text{lab}}(t),
$$
with
$$
R_z(\theta)=
\begin{bmatrix}
\cos\theta & -\sin\theta & 0\\\\
\sin\theta & \cos\theta  & 0\\\\
0 & 0 & 1
\end{bmatrix}.
$$

Set $\theta=\omega_{\mathrm{ref}}t$ and $\psi=\omega_{\mathrm{rf}}t+\phi$. Then
$$
\begin{aligned}
B_x'(t) &= \cos\theta\;B_1\cos\psi + \sin\theta\;B_1\sin\psi \;=\; B_1\cos(\psi-\theta),\\\\
B_y'(t) &= -\sin\theta\;B_1\cos\psi + \cos\theta\;B_1\sin\psi \;=\; B_1\sin(\psi-\theta),\\\\
B_z'(t) &= B_0. 
\end{aligned}
$$

**Key observation:** in the primed frame the transverse RF rotates at the **difference frequency** $\omega_{\mathrm{rf}}-\omega_{\mathrm{ref}}$:
$$
\boxed{ ~~ \mathbf B'(t) = 
\begin{bmatrix}
B_1\cos\big((\omega_{\mathrm{rf}}-\omega_{\mathrm{ref}})t + \phi\big)\\\\
B_1\sin\big((\omega_{\mathrm{rf}}-\omega_{\mathrm{ref}})t + \phi\big)\\\\
B_0
\end{bmatrix}. ~~ }
$$

---

## 3) Pick the co-rotating frame ($\omega_{\mathrm{ref}}=\omega_{\mathrm{rf}}$)

Choosing the frame to **co-rotate with the RF** kills the time dependence in the transverse components:
$$
\boxed{ ~~ \omega_{\mathrm{ref}}=\omega_{\mathrm{rf}} ~~ } \;\Longrightarrow\;
\boxed{ ~~ \mathbf B' =
\begin{bmatrix}
B_1\cos\phi\\\\
B_1\sin\phi\\\\
B_0
\end{bmatrix} \quad \text{(constant in time).} ~~ }
$$

**Interpretation:** all the fast RF rotation has been absorbed by the coordinate axes. In the primed frame, the RF is just a *static* transverse field whose **in-plane direction** is set by $\phi$.

---

## 4) Effective field in the rotating frame

When expressing the **equations of motion** in the primed frame, the rotation of axes introduces a $z$-directed “fictitious” field of magnitude $\omega_{\mathrm{ref}}/\gamma$. That gives the **effective field**
$$
\boxed{ ~~
\mathbf B'_{\mathrm{eff}} \;=\; \mathbf B' \;-\; \frac{\omega_{\mathrm{ref}}}{\gamma}\,\hat{\mathbf z}
\;=\;
\begin{bmatrix}
B_1\cos\phi\\\\
B_1\sin\phi\\\\
B_0-\omega_{\mathrm{ref}}/\gamma
\end{bmatrix}. ~~ }
$$

With the **detuning** defined as
$$
\boxed{ ~~ \Delta\omega \;\equiv\; \gamma B_0 - \omega_{\mathrm{rf}} ~~ }
$$
and for the co-rotating choice $\omega_{\mathrm{ref}}=\omega_{\mathrm{rf}}$, we get
$$
\boxed{ ~~ \mathbf B'_{\mathrm{eff}} =
\begin{bmatrix}
B_1\cos\phi\\\\
B_1\sin\phi\\\\
\Delta\omega/\gamma
\end{bmatrix}. ~~ }
$$

**On resonance** ($\Delta\omega=0$): the effective field lies entirely in the transverse plane along the angle $\phi$.  
**Off resonance** ($\Delta\omega\neq 0$): the effective field tilts out of plane by angle
$$
\theta \;=\; \arctan\!\left(\frac{\gamma B_1}{\Delta\omega}\right),
\qquad
\Omega \;=\; \gamma\|\mathbf B'_{\mathrm{eff}}\| \;=\; \sqrt{(\gamma B_1)^2 + (\Delta\omega)^2}.
$$

---

## 5) What this buys you

- **Conceptual:** In the rotating frame the drive is *static*, so dynamics are rotations about a fixed axis $\mathbf B'_{\mathrm{eff}}$.  
- **Computational:** For a constant-amplitude/phase pulse you can use a single rotation matrix.  
- **Design knobs:** $\phi$ steers the **in-plane axis**; $\Delta\omega$ (i.e., mistuning or $B_0$ inhomogeneity) sets the **tilt**.
