```python
import scipy.io as sio 
import numpy as np 
import matplotlib.pyplot as plt 

%matplotlib inline
import numpy as np
from ipywidgets import interact, FloatSlider, Checkbox
from IPython.display import HTML, display
import matplotlib.pyplot as plt

slc = 90
df = sio.loadmat('/Users/nmickevicius/dev/MCW_BIOP_03-238_MRI/data/brain_relaxometry_3T.mat')
T1 = df['T1'][:,:,slc]
M0 = df['PD'][:,:,slc]
T2 = df['T2'][:,:,slc]
T1[T1 == 0] = 1.0

def plot_all(TI=1.0):
    signal = M0 * (1 - 2*np.exp(-TI/T1))

    fig = plt.figure(figsize=(10.0, 10.0))

    ax1 = fig.add_subplot(3,1,1)
    ax1.imshow(np.abs(signal), vmin=0.0, vmax=1.0, cmap='gray')
    plt.show()

# Widgets / Controls
interact(
    plot_all,
    TI=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.05, description="Inversion Time [s]"),
);
```
