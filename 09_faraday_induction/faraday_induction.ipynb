# Faraday Induction — Full Explanation

Faraday induction—more formally called **Faraday’s Law of Electromagnetic Induction**—is one of the key principles of electromagnetism. It explains how a changing magnetic field can produce an electric current.

---

### The Basic Idea
- **Magnetic fields and electric currents are linked.**
- If the magnetic field through a wire loop changes with time, the wire experiences an induced voltage (also called an **electromotive force, EMF**).
- This induced voltage can drive a current if the loop is part of a closed circuit.

---

### Faraday’s Law
Mathematically, Faraday’s Law is written as:

$$\mathcal{E} = -\frac{d\Phi_B}{dt}$$

Where:
- $\mathcal{E}$ = induced EMF (voltage)  
- $\Phi_B$ = magnetic flux (the amount of magnetic field passing through the loop)  
- $\dfrac{d\Phi_B}{dt}$ = the rate of change of magnetic flux

---

### What is Magnetic Flux?
Magnetic flux measures how much of the magnetic field passes through an area:
$$\Phi_B = B\,A\,\cos(\theta)$$

- $B$ = magnetic field strength  
- $A$ = area of the loop  
- $\theta$ = angle between the field and the loop’s surface **normal** (perpendicular)

Flux increases if:
- The magnetic field gets stronger,
- The loop area changes,
- The angle between the loop and the field changes.

---

### Lenz’s Law (Direction of Induced Current)
The negative sign in Faraday’s Law is explained by **Lenz’s Law**:
- The induced current always flows in a direction that **opposes the change** in magnetic flux.
- For example, if magnetic flux into the page is increasing, the induced current creates a field pointing out of the page to resist that increase.
